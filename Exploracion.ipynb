```python
from pathlib import Path

docs_path = Path("docs-raw-texts")
queries_path = Path("queries-raw-texts")

docs = list(docs_path.iterdir())
queries = list(queries_path.iterdir())

print("Documentos:", len(docs))
print("Queries:", len(queries))

print("\nPrimeros documentos:")
print(docs[:5])

print("\nPrimeras queries:")
print(queries[:5])
```

```python
for doc in docs[:10]:
    print(doc.name)
```

```python
doc = docs[5]

with open(doc, "r", encoding="utf-8") as f:
    contenido = f.read()

print(contenido[:3000])
```

```python
import xml.etree.ElementTree as ET

tree = ET.parse(docs[0])
root = tree.getroot()

print(root.tag)

for child in root:
    print(child.tag)
```

```python
raw = root.find("raw")

print(raw.text)
```

```python
for elem in root.iter():
    print(elem.tag, elem.attrib)
```
